# 05: Modern Depthwise CNN (ConvNeXt-Tiny) & Temporal CRNN
**Architectures:**
1. **ConvNeXt-Tiny:** Modernized pure convolution (7x7 depthwise separable filters, inverted bottlenecks, LayerNorm).
2. **CRNN:** Bioacoustic CNN + Bidirectional GRU + Temporal Attention Pooling.


```python
# Cell 1: Environment & Paths
import json
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

PROJECT_ROOT = Path('.').resolve().parent if Path('.').resolve().name == 'code' else Path('.').resolve()
MODELS_DIR = PROJECT_ROOT / 'model_artifacts'
print('Artifacts Directory:', MODELS_DIR)
```

```python
# Cell 2: ConvNeXt-Tiny Diagnostics & Holdout Metrics
with open(MODELS_DIR / 'convnext_tiny' / 'final_evaluation_report.json') as f: cnx_rep = json.load(f)
print('=== ConvNeXt-Tiny Top-3 Multi-Window Ensemble ===')
print(f'Top-1 Test Accuracy : {cnx_rep["test_accuracy_top1"]*100:.2f}%')
print(f'Top-3 Test Accuracy : {cnx_rep["test_accuracy_top3"]*100:.2f}%')
print(f'Test Macro F1       : {cnx_rep["test_macro_f1"]*100:.2f}%')

display(Image(filename=str(MODELS_DIR / 'convnext_tiny' / 'final_confusion_matrix.png')))
```

```python
# Cell 3: CRNN Diagnostics & Holdout Metrics
with open(MODELS_DIR / 'crnn' / 'final_evaluation_report.json') as f: crnn_rep = json.load(f)
print('=== Bioacoustic CRNN (CNN + Bi-GRU) Top-3 Multi-Window Ensemble ===')
print(f'Top-1 Test Accuracy : {crnn_rep["test_accuracy_top1"]*100:.2f}%')
print(f'Top-3 Test Accuracy : {crnn_rep["test_accuracy_top3"]*100:.2f}%')
print(f'Test Macro F1       : {crnn_rep["test_macro_f1"]*100:.2f}%')

display(Image(filename=str(MODELS_DIR / 'crnn' / 'final_confusion_matrix.png')))
```
